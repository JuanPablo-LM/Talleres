## **4. Cargue los datos en el notebook.**

```python
# Llaves de identificación
llaves_hogar = ["DIRECTORIO", "SECUENCIA_P"]
llaves_persona = ["DIRECTORIO", "SECUENCIA_P", "ORDEN"]

# Diccionario para guardar las bases de cada mes
bases_mensuales = {}
```

```python
# Recorrer las carpetas de cada mes
for carpeta in ruta.iterdir():

    caracteristicas = None
    hogar = None
    fuerza = None
    ocupados = None

    # Buscar y cargar los archivos
    for archivo in carpeta.glob("*.CSV"):

        nombre = archivo.name.lower()

        if "caracter" in nombre:
            caracteristicas = pd.read_csv(archivo, sep=";", decimal=',', encoding = "latin-1", dtype = str)

        elif "datos del hogar" in nombre:
            hogar = pd.read_csv(archivo, sep=";", decimal=',', encoding = "latin-1", dtype = str)

        elif "fuerza" in nombre:
            fuerza = pd.read_csv(archivo, sep=";", decimal=',', encoding = "latin-1", dtype = str)

        elif "ocupados" in nombre and "no ocupados" not in nombre:
            ocupados = pd.read_csv(archivo, sep=";", decimal=',', encoding = "latin-1", dtype = str)



    # Unir información del hogar
    columnas_hogar = llaves_hogar + [col for col in hogar.columns if col not in caracteristicas.columns]
    base_mes = caracteristicas.merge(hogar[columnas_hogar], on=llaves_hogar, how="left")


    # Unir fuerza de trabajo
    columnas_fuerza = llaves_persona + [col for col in fuerza.columns if col not in base_mes.columns]
    base_mes = base_mes.merge(fuerza[columnas_fuerza], on=llaves_persona, how="left")


    # Unir ocupados
    columnas_ocupados = llaves_persona + [col for col in ocupados.columns if col not in base_mes.columns]
    base_mes = base_mes.merge(ocupados[columnas_ocupados], on=llaves_persona, how="left")


    # Guardar la base resultante
    bases_mensuales[carpeta.name] = base_mes
```

## **5. Para cada mes de 2026, construya una base de datos uniendo la información**

```python
# Bases unificadas por mes
for mes, df in bases_mensuales.items():
    print("\n", mes)
    print("Dimensiones:", df.shape)
    display(df.head())
```

## **6. Concatene las bases de todos los meses**

```python
# Concatenar las bases de todos los meses
base_geih_2026 = pd.concat(bases_mensuales.values(), ignore_index=True)

# Ver las dimensiones y las primeras filas
print("Dimensiones:", base_geih_2026.shape)
display(base_geih_2026.head())
```

## **7. Dataframe 2025-01-14 Correlativa CIIU Rev4.xlsx**

```python
# Leer la tabla de correspondencia CIIU
archivo_ciiu = ruta.parent / "2025-01-14 Correlativa CIIU Rev4.xlsx"
ciiu = pd.read_excel(archivo_ciiu, sheet_name="CIIU 2022", dtype=str)

# Seleccionar código y descripción
ciiu = ciiu[["RAMA4D_R4", "DESCRIPCION_CIIU"]]

# Agregar la descripción CIIU a la base
base_geih_2026 = base_geih_2026.merge(ciiu, on="RAMA4D_R4", how="left")

# Verificar las dimensiones
print("Dimensiones:", base_geih_2026.shape)

# Verificar 10 observaciones aleatorias con código y descripción CIIU
print("\n10 observaciones aleatorias:")
display(
    base_geih_2026[
        ["RAMA4D_R4", "DESCRIPCION_CIIU"]
    ].dropna().sample(10, random_state=42)
)
```

## **8. Dataframe 2025-01-14 Divipola.xlsx**

```python
# Leer la tabla de departamentos de DIVIPOLA
archivo_divipola = ruta.parent / "2025-01-14 DIVIPOLA.xlsx"
divipola = pd.read_excel(archivo_divipola, sheet_name="Departamentos", header=9, dtype=str)

# Seleccionar código y nombre del departamento
divipola = divipola[["Código", "Nombre"]]

# Renombrar las columnas para hacer el cruce con la base
divipola = divipola.rename(columns={"Código": "DPTO", "Nombre": "NOMBRE_DPTO"})

# Agregar el nombre del departamento a la base
base_geih_2026 = base_geih_2026.merge(divipola, on="DPTO", how="left")

# Verificar las dimensiones
print("Dimensiones:", base_geih_2026.shape)

# Verificar 10 departamentos aleatorios
print("\n10 departamentos aleatorios:")
display(
    base_geih_2026[
        ["DPTO", "NOMBRE_DPTO"]
    ].dropna().drop_duplicates().sample(10, random_state=42)
)
```

## **9. Guarde la base de la GEIH 2026 en un archivo csv.**

```python
# Guardar la base GEIH 2026 como CSV
archivo_salida = ruta.parent / "Base_GEIH_2026.csv"
base_geih_2026.to_csv(archivo_salida, index=False, encoding="utf-8")

# Verificar que el archivo fue guardado
print("Base guardada en:", archivo_salida)
print("Dimensiones:", base_geih_2026.shape)
```
